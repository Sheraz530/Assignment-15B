```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Noman Traders\Downloads\student_tea_snacks.csv")
df.info()
```

```python
#Q1: What is the most preferred tea type?
#Answer: Green
df.groupby("tea_type").count()
```

```python
#Q2: What is the average number of cups per gender?
#Answer:
df.groupby("gender")["cups_per_day"].mean()
```

```python
#Q3: Make a bar chart of snack choices.
#Answer:
import matplotlib.pyplot as plt
import seaborn as sns
sns.barplot(data=df, x='cups_per_day', y='snack_choice', orient='h', estimator='mean')
plt.xlabel('Average cups per day')
plt.ylabel('snack_choice')
plt.title('Snack choices')
plt.show()
```

```python
#Q4: Compare tea preference by age group.
#Answer:
df.groupby("age_group")["prefers_tea"].value_counts()
```

```python
#Q5: Count the number of students who don't drink tea
#Answer:49 Students
df.groupby("prefers_tea").count()
```
